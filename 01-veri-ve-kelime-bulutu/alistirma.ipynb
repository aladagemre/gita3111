**Soru 3** — Hatayı bul: hangi satır, neden? `kütüphane` yazmalı.

```python
def en_uzun_kelime(kelimeler):
    en_uzun = ""
    for kelime in kelimeler:
        if len(kelime) > len(en_uzun):
            en_uzun = kelime
        return en_uzun

print(en_uzun_kelime(["ev", "okul", "kütüphane", "yol"]))
```

**Soru 4** — Hatayı bul: `kahve 2` yazmalı. Hata mesajı yok, ama sonuç yanlış.

```python
from collections import Counter

metin = "kahve çay kahve"
sayac = Counter(metin)
for kelime, adet in sayac.most_common(1):
    print(kelime, adet)
```
