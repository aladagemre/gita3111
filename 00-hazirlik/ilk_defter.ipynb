## İlk grafiğin

Aşağıdaki iki listeyi ve başlığı **kendi verinle** değiştir: bir haftada kaç saat çizim
yaptığın, en çok kullandığın beş rengi kaç kez kullandığın… Sonra hücreyi yeniden çalıştır.

```python
import matplotlib.pyplot as plt

gunler = ["Pzt", "Sal", "Çar", "Per", "Cum"]
saatler = [3, 7, 2, 8, 5]

plt.bar(gunler, saatler)
plt.title("Bu hafta kaç saat çizim yaptım?")
plt.show()
```
