**Soru 1 — Tarifi yeniden yaz.** Aşağıdaki aracın tarifi model için işe yaramaz. Bir brif gibi yeniden yaz: araç ne

İpucu: `olcu_ver("Instagram hikâyesi")` yazıp çalıştır. Model `bicim`'e böyle bir şey yazarsa ne olur? Tarifin
buna engel olmalı.

```python
def olcu_ver(bicim: str):
    """Ölçü."""
    olculer = {"gonderi": "1080 x 1350 piksel", "hikaye": "1080 x 1920 piksel", "a3": "297 x 420 mm"}
    return olculer[bicim]

tarifi_goster(olcu_ver)
```
