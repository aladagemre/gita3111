Sayılar tırnak içinde, çünkü CSV'den okunan her şey metindir.

```python
kayitlar = [
    {"ad": "kırmızı", "r": "230", "g": "57", "b": "70", "etiket": "enerjik"},
    {"ad": "orta mavi", "r": "69", "g": "123", "b": "157", "etiket": "sakin"},
    {"ad": "lacivert", "r": "29", "g": "53", "b": "87", "etiket": "ciddi"},
    {"ad": "şeftali", "r": "244", "g": "162", "b": "97", "etiket": "enerjik"},
    {"ad": "kırık beyaz", "r": "241", "g": "250", "b": "238", "etiket": "sakin"},
    {"ad": "hardal", "r": "233", "g": "196", "b": "106", "etiket": "enerjik"},
]
print(len(kayitlar))
```

```python
X = []
y = []
for kayit in kayitlar:
    r = int(kayit["r"])
    g = int(kayit["g"])
    b = int(kayit["b"])
    X.append([r, g, b])
    y.append(kayit["etiket"])

print(X[0], y[0])
```

**2)** Yalnız enerjik renkler. İki liste de 3 elemanlı olmalı: `3 3`.

```python
adlar = []
etiketler = []
for kayit in kayitlar:
    if kayit["etiket"] == "enerjik":
        adlar.append(kayit["ad"])
    etiketler.append(kayit["etiket"])

print(len(adlar), len(etiketler))
```

**3)** Eğitim etiketleri. `y_egitim` sayılardan değil, etiketlerden oluşmalı: `['enerjik', 'enerjik', 'sakin']`.

```python
from sklearn.model_selection import train_test_split

X_egitim, y_egitim, X_test, y_test = train_test_split(
    X, y, test_size=0.5, random_state=0
)
print(y_egitim)
```

**4)** Renklerin kaçta kaçı enerjik? `0.5` yazmalı.

```python
enerjik = 0
for etiket in y:
    if etiket == "Enerjik":
        enerjik = enerjik + 1

oran = enerjik / len(y)
print(oran)
```

**5)** Doğruluk, modelin **görmediği** renklerde ölçülmeli. `0.3333333333333333` yazmalı.

```python
from sklearn.neighbors import KNeighborsClassifier

X_egitim, X_test, y_egitim, y_test = train_test_split(
    X, y, test_size=0.5, random_state=0
)
model = KNeighborsClassifier(n_neighbors=1)
model.fit(X_egitim, y_egitim)
print(model.score(X_egitim, y_egitim))
```
