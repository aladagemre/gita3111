Önce bu hücreyi çalıştır; aşağıdaki hücreler bunu kullanıyor:

```python
import json
from pathlib import Path

stiller = ["kolaj", "neon"]
renkler = ["mavi", "sari", "mor"]
stil_yonergeleri = {"kolaj": "paper collage style", "neon": "neon sign style"}
kartlar = [
    {"stil": "kolaj", "renk": "mavi", "dosya": "kolaj_mavi.jpg"},
    {"stil": "kolaj", "renk": "sari", "dosya": "kolaj_sari.jpg"},
    {"stil": "neon", "renk": "mavi", "dosya": "neon_mavi.jpg"},
]
print(len(stiller), len(renkler), len(kartlar))
```

## Bozuk kodlar

Her hücrenin üstünde ne yapması gerektiği yazıyor. Çalıştır, hatayı ya da yanlış çıktıyı gör, düzelt.
Hücreler birbirinden bağımsız.

**1)** Bütün kombinasyonları yazmalı: 6 satır (`kolaj mavi`, `kolaj sari`, ... `neon mor`).

```python
for stil in stiller:
    for renk in renkler:
    print(stil, renk)
```

**2)** Bütün kombinasyonları yazmalı: 6 satır. Hata vermiyor ama yalnızca 2 satır yazıyor.

```python
for stil in stiller:
    for renk in renkler:
        sonuc = stil + " " + renk
    print(sonuc)
```

**3)** Her kartın dosya adını yazmalı: `kolaj_mavi.jpg`, `kolaj_sari.jpg`, `neon_mavi.jpg`.

```python
for kart in kartlar:
    print(kart[2])
```

**4)** En fazla 2 kart işlemeli: `kolaj_mavi.jpg`, `kolaj_sari.jpg`, sonra `Sınıra geldim`. Hata vermiyor ama 3 kartı da işliyor.

```python
MAKS_URETIM = 2
uretilen = 0
for kart in kartlar:
    if uretilen > MAKS_URETIM:
        print("Sınıra geldim")
        break
    print(kart["dosya"])
    uretilen = uretilen + 1
```

**5)** Kartları `kartlar.json` dosyasına yazmalı, sonra geri okuyup kart sayısını yazmalı: `3`.

```python
with open("kartlar.json", "w", encoding="utf-8") as dosya:
    json.dump(dosya, kartlar)

with open("kartlar.json", encoding="utf-8") as dosya:
    okunan = json.load(dosya)
print(len(okunan))
```
