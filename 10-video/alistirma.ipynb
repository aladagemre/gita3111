**1)** En fazla 6 kez, 2 saniye arayla sormalı: üç kez `Henüz bitmedi`, sonra `Hazır`. Hata vermiyor ama hemen `Zaman aşımı` diyor.

```python
yanit = sahte_baslat(govde)
is_no = yanit["is_no"]

for deneme in range(6):
    durum = sahte_durum(is_no)
    if durum["bitti"]:
        print("Hazır")
        break
    print("Henüz bitmedi")

if not durum["bitti"]:
    print("Zaman aşımı")
```
