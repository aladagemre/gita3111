```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

veri = np.load("veri/harfler.npz")
resimler = veri["R"] / 255
fontlar = list(veri["fontlar"])
turler = list(veri["turler"])
print(resimler.shape)
```

## Bozuk kodlar

Her hücrenin üstünde, hücrenin ne yapması gerektiği yazıyor. Çalıştır, hatayı ya da yanlış çıktıyı gör,
hücreyi düzelt, yeniden çalıştır.

**1)** İlk fontun R harfini çizmeli.

```python
plt.imshow(resimler[0], cmap="gray_r")
plt.show()
```

**2)** R'leri 2 sayıya sıkıştırıp ilk fontun kodunu yazmalı (iki sayı).

```python
pca = PCA(n_components=2)
kodlar = pca.transform(resimler)
pca.fit(resimler)
print(kodlar[0])
```

**3)** 2. hücreyi düzelttikten sonra çalıştır. Haritanın ortasındaki noktadan bir R üretip çizmeli.

```python
yeni = pca.inverse_transform([0, 0])
plt.imshow(yeni[0].reshape(28, 28), cmap="gray_r")
plt.show()
```

**4)** İki noktanın tam ortasını yazmalı: `[4. 3.]`.

```python
bas = np.array([2, 2])
son = np.array([6, 4])
oran = 0.5
ara = bas + oran * son - bas
print(ara)
```

**5)** İlk üç fontun R'sini yan yana üç kutuya çizmeli.

```python
sira = 0
for resim in resimler[:3]:
    plt.subplot(1, 3, sira)
    plt.imshow(resim.reshape(28, 28), cmap="gray_r")
    sira = sira + 1
plt.show()
```
