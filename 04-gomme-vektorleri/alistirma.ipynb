```python
kahve = [0.6, 0.4]
cay = [0.5, -0.4]
buz = [-0.9, 0.1]
deniz = [-0.5, -0.5]
ates = [0.9, 0.9]
kar = [-0.7, -0.2]

harita = {
    "kahve": kahve,
    "çay": cay,
    "buz": buz,
    "deniz": deniz,
    "ateş": ates,
    "kar": kar,
}
print(len(harita))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def benzerlik(a, b):
    sonuc = cosine_similarity([a], [b])
    satir = sonuc[0]
    return satir[0]

print(benzerlik(kahve, cay))
```

Bu hücre her kelimenin kahveye benzerliğini `skorlar` sözlüğüne koyar:

```python
skorlar = {}
for kelime in harita:
    vektor = harita[kelime]
    skorlar[kelime] = benzerlik(kahve, vektor)

print(len(skorlar))
```

```python
sonuc = cosine_similarity(kahve, buz)
satir = sonuc[0]
print(satir[0])
```

**2)** Kahveye en yakın **3** kelime, kahvenin kendisinin altında. Dört satır yazmalı:
`kahve 1.0`, `ateş 0.98`, `çay 0.3`, `buz -0.77`.

```python
from collections import Counter

sirali = Counter(skorlar)
for kelime, skor in sirali.most_common(3):
    print(kelime, round(skor, 2))
```

**3)** (Adım 7'deki PCA'yı gördükten sonra çöz.) Kahvenin haritadaki iki sayısı. Hata vermeden `[0.70881312 0.01901953]` yazmalı.

```python
from sklearn.decomposition import PCA

liste = [kahve, cay, buz, deniz, ates, kar]
pca = PCA(n_components=2)
noktalar = pca.transform([kahve])
print(noktalar[0])
```
