Önce hazırlık: takvimi oku ve yardımcı fonksiyonları al.

````python
import csv
from uretim import gonderi_yap, onay_sayfasi, goster

with open("veri/takvim.csv", encoding="utf-8-sig") as dosya:
    okuyucu = csv.DictReader(dosya)
    takvim = list(okuyucu)
````

Pazartesi'nin hazır görselinden bir gönderi yap; başlık takvimden gelsin.
Çalıştırmadan önce düşün: bu hücre modele kaç istek gönderir?

## 2. Bozuk kod: beş gönderi, tek dosya

Döngü beş gönderi yaptı. Çalıştır, sonra `alistirma_ciktilari/` klasörüne bak: kaç gönderi dosyası var, hangi günün
gönderisi görünüyor? Neden? Kodu düzelt: her gün kendi dosyasına yazılsın.

````python
for satir in takvim:
    gun = satir["gun"]
    gonderi_yap(f"veri/hazir/{gun}.jpg", satir["baslik"], "gonderi", "alistirma_ciktilari/gonderi.jpg")

goster("alistirma_ciktilari/gonderi.jpg")
````
