çalışır: **renk · yazı tipi · açılış saati**. Model yok; yönergeleri yalnızca kurup yazdırıyoruz.

Önce bu iki hücreyi çalıştır; aşağıdaki hücreler bunları kullanıyor:

````python
from collections import Counter
from sklearn.metrics.pairwise import cosine_similarity

def benzerlik(a, b):
    sonuc = cosine_similarity([a], [b])
    satir = sonuc[0]
    return satir[0]

kilavuz_vektorleri = {
    "Ana renkler mürekkep mavisi ve kâğıt beyazıdır.": [0.9, 0.1, 0.0],
    "Kırmızı ve turuncu hiçbir tasarımda kullanılmaz.": [0.8, 0.2, 0.0],
    "Başlıklarda serifli bir yazı tipi kullanılır.": [0.1, 0.9, 0.0],
    "Menü kartları beyaz zemine mavi mürekkeple basılır.": [0.6, 0.5, 0.0],
    "Kafe hafta içi 08.00–23.00 arası açıktır.": [0.0, 0.0, 1.0],
    "Sınav haftalarında kapanış gece yarısına uzar.": [0.0, 0.2, 0.8],
}
talimat = "Yalnızca aşağıdaki bilgilere dayanarak cevap ver. Bilgilerde cevap yoksa yalnızca 'Bilmiyorum' yaz."
print("Hazır:", len(kilavuz_vektorleri), "paragraf")
````

Bu hücre, "Afişte hangi renkleri kullanayım?" sorusu (`[1.0, 0.0, 0.0]`) için her paragrafın puanını
`renk_skorlari` sözlüğüne koyar:

````python
renk_sorusu = [1.0, 0.0, 0.0]

renk_skorlari = {}
for paragraf in kilavuz_vektorleri:
    vektor = kilavuz_vektorleri[paragraf]
    renk_skorlari[paragraf] = benzerlik(renk_sorusu, vektor)

print(len(renk_skorlari))
````

## Bozuk kodlar

Her hücrenin üstünde ne yapması gerektiği yazıyor. Çalıştır, hatayı ya da yanlış çıktıyı gör, düzelt.
Hücreler birbirinden bağımsız.

**1)** Sayfa Kahve'nin kılavuzunu okuyup paragraf sayısını yazmalı: `14`.

````python
paragraflar = []
with open("kilavuz.txt", encoding="utf-8") as dosya:
    for satir in dosya:
        paragraf = satir.strip()
        if paragraf != "":
            paragraflar.append(paragraf)

print(len(paragraflar))
````

**2)** Renk sorusuna en yakın üç paragrafı puanlarıyla yazmalı: `0.99 Ana renkler ...`, `0.97 Kırmızı ...`,
`0.77 Menü kartları ...`. Hata vermiyor ama üç satırda da puan `1.0`.

````python
soru_vektoru = [1.0, 0.0, 0.0]

skorlar = {}
for paragraf in kilavuz_vektorleri:
    vektor = kilavuz_vektorleri[paragraf]
    skorlar[paragraf] = benzerlik(soru_vektoru, soru_vektoru)

sirali = Counter(skorlar)
for paragraf, skor in sirali.most_common(3):
    print(round(skor, 2), paragraf)
````

**3)** Renk sorusuna en yakın iki paragrafı tek bir metinde toplamalı; iki satır yazmalı: `Ana renkler ...` ve `Kırmızı ...`.
Hata vermiyor ama tek satır yazıyor.

````python
sirali = Counter(renk_skorlari)
bilgiler = ""
for paragraf, skor in sirali.most_common(2):
    bilgiler = paragraf + "\n"

print(bilgiler)
````

**4)** Her paragrafın başına puanını yazmalı: `0.99 Ana renkler ...` ve `0.97 Kırmızı ...`.

````python
sirali = Counter(renk_skorlari)
bilgiler = ""
for paragraf, skor in sirali.most_common(2):
    bilgiler = bilgiler + round(skor, 2) + " " + paragraf + "\n"

print(bilgiler)
````

**5)** "Sınav haftasında kaçta kapanıyorsunuz?" sorusuna (`[0.0, 0.0, 1.0]`) en yakın iki paragrafı yazmalı:
`Kafe hafta içi ...` ve `Sınav haftalarında ...`. Hata vermiyor ama `None` yazıyor.

````python
def elle_ara(soru_vektoru):
    skorlar = {}
    for paragraf in kilavuz_vektorleri:
        vektor = kilavuz_vektorleri[paragraf]
        skorlar[paragraf] = benzerlik(soru_vektoru, vektor)
    sirali = Counter(skorlar)
    bilgiler = ""
    for paragraf, skor in sirali.most_common(2):
        bilgiler = bilgiler + paragraf + "\n"

bulunan = elle_ara([0.0, 0.0, 1.0])
print(bulunan)
````
