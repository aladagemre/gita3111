## Soru 2 — Doldur: el yapımı vektörlerle puan

Konu 4'teki gibi elle verilmiş küçük vektörler. Her metnin üç sayısı var: ne kadar **renkten**, ne kadar **yazıdan**,
ne kadar **zamandan** söz ediyor (0 hiç, 1 tamamen). Soru: "Başlığı turuncu yapabilir miyim?"

```python
from sklearn.metrics.pairwise import cosine_similarity

def benzerlik(a, b):
    sonuc = cosine_similarity([a], [b])
    satir = sonuc[0]
    return satir[0]

kucuk_hafiza = {
    "Kırmızı ve turuncu kullanılmaz.": [0.9, 0.1, 0.0],
    "Başlıkta serifli yazı tipi kullanılır.": [0.1, 0.9, 0.0],
    "Matbaa cuma 17.00'den sonra gelen işi pazartesi basar.": [0.0, 0.1, 0.9],
    "Açık sarı yazıyı açık zemine koyma.": [0.6, 0.5, 0.0],
}
soru_vektoru = [0.8, 0.3, 0.0]
```

**Tahmin et:** en yüksek puanı hangisi alır, en düşüğü hangisi? Sonra `BOSLUK`'u doldur: her metnin vektörüyle
sorunun vektörünü karşılaştır.

```python
puanlar = {}
for metin in kucuk_hafiza:
    vektor = kucuk_hafiza[metin]
    puanlar[metin] = benzerlik(soru_vektoru, BOSLUK)
    print(puanlar[metin], metin)
```

## Soru 3 — Doldur: en yüksek puanı bul

`hafizada_ara` puanları büyükten küçüğe diziyordu. En büyüğünü bulmak için puanları tek tek gezip o ana kadarki en
büyükle karşılaştırmak yeter. `BOSLUK`'u doldur. Doğru çıktı: `Kırmızı ve turuncu kullanılmaz.`

```python
en_iyi_not = ""
en_iyi_puan = -1
for metin in puanlar:
    puan = puanlar[metin]
    if puan > BOSLUK:
        en_iyi_puan = puan
        en_iyi_not = metin
print(en_iyi_not)
```

## Soru 5 — Düzelt: üç haftalık hafıza bir satırda gitti

Gerçek dosyayı bozmamak için önce notların bir kopyasını `ciktilar/` içine alıyoruz.
