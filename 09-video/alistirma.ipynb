## Kendi başına

`BOSLUK` yazan yerleri doldur. Doldurmadan çalıştırırsan hata alırsın (bazılarında `NameError: name 'BOSLUK' is not defined`,
bazılarında başka bir hata). Her sorunun üstünde doğru çıktı yazılı.

**Soru 1 — Doldur:** 10 saniyelik 1080p klibin fiyatı: `10 saniye, 1080p: 1.52 USD`.

````python
sure = 10
fiyat = fiyatlar[BOSLUK]
ucret = round(sure * fiyat, 2)
print(sure, "saniye, 1080p:", ucret, "USD")
````

**Soru 2 — Doldur:** 6 saniyelik bir klibin her çözünürlükteki fiyatı, dört satır: `360p : 0.2 USD`, `720p : 0.61 USD`,
`1080p : 0.91 USD`, `4K : 1.82 USD`.

````python
sure = 6
for cozunurluk in BOSLUK:
    fiyat = fiyatlar[cozunurluk]
    ucret = round(sure * fiyat, 2)
    print(cozunurluk, ":", ucret, "USD")
````

**Soru 3 — Doldur:** bütçe 1 dolar; 1080p'de bütçeye sığan süreleri yazmalı: `4 saniye sığıyor` ve `6 saniye sığıyor`.

````python
butce = 1.0
fiyat = fiyatlar["1080p"]
for sure in [4, 6, 8, 10]:
    ucret = round(sure * fiyat, 2)
    if ucret BOSLUK butce:
        print(sure, "saniye sığıyor")
````

**Soru 4 — Doldur:** iki kaydı günlüğe yazmalı: `Yazıldı`. (Soru 5 bu dosyayı okuyor.)

````python
kayitlar = [
    {"yonerge": "rain on a cafe window, slow pan", "sure": 8, "cozunurluk": "720p"},
    {"yonerge": "pages of a book turning in the wind", "sure": 6, "cozunurluk": "1080p"},
]
with open("alistirma_gunlugu.json", "w", encoding="utf-8") as dosya:
    json.BOSLUK(kayitlar, dosya, ensure_ascii=False, indent=2)
print("Yazıldı")
````

**Soru 5 — Doldur:** günlüğü okuyup kliplerin toplam süresini yazmalı: `Toplam süre: 14 saniye`.

````python
with open("alistirma_gunlugu.json", encoding="utf-8") as dosya:
    okunan = json.load(dosya)

toplam = 0
for kayit in okunan:
    toplam = toplam + kayit[BOSLUK]
print("Toplam süre:", toplam, "saniye")
````
