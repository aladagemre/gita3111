## 2. Bozuk kod — yanlış liste

Deniz günü `gun_adi` olmadan, kendisi hesaplamak istemiş. Kodu 14 Kasım 2026 için "Cuma" diyor: afişteki hatanın
aynısı. Çalıştır, sorunu bul, listeyi düzelt.

İpucu: `weekday()` Pazartesi'ye 0, Salı'ya 1, ..., Pazar'a 6 der.

```python
import datetime

gunler = ["Pazar", "Pazartesi", "Salı", "Çarşamba", "Perşembe", "Cuma", "Cumartesi"]
gun = datetime.date(2026, 11, 14)
sira = gun.weekday()
print(sira, gunler[sira])
```

## 3. Bozuk kod — eşik ters

`veri/afis_temiz.jpg` aynı afişin düzeltilmiş hâli. Kod onda zor okunan renkleri arıyor ama mürekkep mavisini
(kontrastı 10'un üstünde) "zor okunur" diye işaretliyor. Neden? Tek bir işareti değiştirerek düzelt.

Düzelttikten sonra listede kalan renge bak: temiz afişi aç. O renk bir yazının mı, bir süsün mü? Rapora "zor okunur"
diye yazar mısın?

```python
renkler = renkleri_cikar("veri/afis_temiz.jpg", 5)
zemin = renkler[0]

for renk in renkler[1:]:
    oran = kontrast(renk, zemin)
    if oran > 4.5:
        print(renk, "zor okunur:", oran)
```
