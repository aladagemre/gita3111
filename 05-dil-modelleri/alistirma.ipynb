Önce bu hücreyi çalıştır; aşağıdaki bütün hücreler bunu kullanıyor:

```python
import tiktoken
from collections import Counter

metin = "kahve sıcak çay sıcak kahve acı kahve sıcak limonata soğuk"
kelimeler = metin.split()
print(len(kelimeler))

kodlayici = tiktoken.get_encoding("o200k_base")
```

## Bozuk kodlar

Her hücrenin üstünde, hücrenin ne yazması gerektiği yazıyor. Çalıştır, hatayı ya da yanlış çıktıyı gör,
hücreyi düzelt, yeniden çalıştır.

**1)** "sessiz" kelimesinin parçalarını alt alta yazmalı: `sess` ve `iz`.

```python
belirtecler = kodlayici.encode("sessiz")
for belirtec in belirtecler:
    print(kodlayici.decode(belirtec))
```

**2)** "kahve"den sonra gelenleri yazmalı: `['sıcak', 'acı', 'sıcak']`.

```python
def sonra_gelenleri_bul(hedef):
    sonra_gelenler = []
    onceki = ""
    for kelime in kelimeler:
        if onceki == hedef:
            sonra_gelenler.append(kelime)
            onceki = kelime
    return sonra_gelenler

print(sonra_gelenleri_bul("kahve"))
```

**3)** Olasılıkların toplamı 1 etmeli: `sıcak 0.67` ve `acı 0.33` yazmalı.

```python
sonra_gelenler = ["sıcak", "acı", "sıcak"]
sayac = Counter(sonra_gelenler)
toplam = len(sayac)
for kelime, adet in sayac.most_common(2):
    olasilik = adet / toplam
    print(kelime, round(olasilik, 2))
```

**4)** 2'yi düzelttikten sonra çalıştır. Fonksiyon en sık gelen kelimeyi vermeli: `sıcak` yazmalı.

```python
def sonraki_kelime(hedef):
    sonra_gelenler = sonra_gelenleri_bul(hedef)
    sayac = Counter(sonra_gelenler)
    en_sik = sayac.most_common(1)
    kelime, adet = en_sik[0]
    return adet

print(sonraki_kelime("kahve"))
```

**5)** 4'ü düzelttikten sonra çalıştır. `kahve sıcak çay sıcak` yazmalı.

```python
kelime = "kahve"
cumle = kelime
for tur in [1, 2, 3]:
    yeni = sonraki_kelime(kelime)
    cumle = cumle + " " + yeni

print(cumle)
```
