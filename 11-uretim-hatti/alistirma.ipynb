Bu defter **anahtar istemez** ve internete çıkmaz: halkalar sahte. `sahte_slogan` ve `sahte_ciz` her çağrıldığında
bir listeye not düşer (`metin_istekleri`, `gorsel_istekleri`); böylece kaç istek gittiğini sayabiliyoruz.
`sahte_ciz` boş yönergeye `None` verir (gerçek model gibi reddeder).

Önce bu hücreyi çalıştır; aşağıdaki hücreler bunu kullanıyor:

```python
import json
from pathlib import Path

klasor = Path("alistirma_ciktilari")
klasor.mkdir(exist_ok=True)

ayarlar = {"konu": "gece açık bir kitapçı", "stil": "linocut print", "slogan_sayisi": 3}
with open(klasor / "ayarlar.json", "w", encoding="utf-8") as dosya:
    json.dump(ayarlar, dosya, ensure_ascii=False)

metin_istekleri = []
gorsel_istekleri = []

def sahte_slogan(konu):
    metin_istekleri.append(konu)
    sira = len(metin_istekleri)
    return f' "{konu} {sira}" '

def sahte_ciz(yonerge):
    gorsel_istekleri.append(yonerge)
    if yonerge == "":
        return None
    return b"sahte gorsel"

print("Hazır")
```

## Bozuk kodlar

Her hücrenin üstünde ne yapması gerektiği yazıyor. Çalıştır, hatayı ya da yanlış çıktıyı gör, düzelt.
Hücreler birbirinden bağımsız.

**1)** Ayarlar dosyasından konuyu okuyup yazmalı: `gece açık bir kitapçı`.

```python
with open(klasor / "ayarlar.json", encoding="utf-8") as dosya:
    okunan = json.load(dosya)

konu = okunan["Konu"]
print(konu)
```

**2)** Üç slogan isteyip hepsini listede toplamalı: `3`. Hata vermiyor ama `1` yazıyor.

```python
for tur in range(3):
    sloganlar = []
    slogan = sahte_slogan("kitapçı")
    sloganlar.append(slogan)

print(len(sloganlar))
```

**3)** Klasörde `gorsel_1.jpg` ve `gorsel_2.jpg` zaten var; yalnızca eksik olan için istek gitmeli: `Giden istek: 1`.
Hata vermiyor ama `Giden istek: 3` yazıyor (her çalıştırmada).

```python
(klasor / "gorsel_1.jpg").write_bytes(b"")   # kurulum satırı, dokunma: varmış gibi boş bir dosya oluşturur
(klasor / "gorsel_2.jpg").write_bytes(b"")   # kurulum satırı, dokunma

once = len(gorsel_istekleri)
sira = 0
for yonerge in ["linocut print, kitap", "linocut print, fener", "linocut print, kedi"]:
    sira = sira + 1
    yol = klasor / f"gorsel_{sira}.jpg"
    resim = sahte_ciz(yonerge)
    print("İstek gitti:", yol)

print("Giden istek:", len(gorsel_istekleri) - once)
```

**4)** Üç yönergeyi sırayla çizmeli; ortadaki boş, onu atlayıp devam etmeli:
`Görsel geldi: 12 bayt`, `Olmadı, atlıyorum`, `Görsel geldi: 12 bayt`.

```python
for yonerge in ["linocut print, kitap", "", "linocut print, fener"]:
    resim = sahte_ciz(yonerge)
    print("Görsel geldi:", len(resim), "bayt")
```

**5)** Her yönerge için bir kayıt tutmalı, olmayanlar dahil: `3` kayıt, ortadakinin durumu `olmadi`. Hata vermiyor ama `2` yazıyor.

```python
kayitlar = []
for yonerge in ["linocut print, kitap", "", "linocut print, fener"]:
    resim = sahte_ciz(yonerge)
    if resim is None:
        continue
    kayit = {"girdi": yonerge, "durum": "uretildi"}
    kayitlar.append(kayit)

print(len(kayitlar))
print(kayitlar)
```
