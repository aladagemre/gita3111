Bu defter **anahtar istemez**. Modelin yanıtını taklit ediyoruz: `veri/fincan.jpg`'yi base64'e çevirip
API'nin gönderdiği biçimde bir sözlüğe koyuyoruz. `.decode()` baytları düz metne çevirir; API'den gelen metin de böyle.

Önce bu hücreyi çalıştır; aşağıdaki bütün hücreler bunu kullanıyor:

```python
import base64
from pathlib import Path
from datetime import datetime

with open("veri/fincan.jpg", "rb") as dosya:
    baytlar = dosya.read()

metin = base64.b64encode(baytlar).decode()
yanit = {"success": True, "result": {"image": metin}}
print(len(baytlar), len(metin))
```

## Kendi başına

`BOSLUK` yazan yerleri doldur. Doldurmadan çalıştırırsan hata alırsın (çoğunda `NameError: name 'BOSLUK' is not defined`;
bazılarında başka bir hata). Her sorunun üstünde doğru çıktı yazılı.

**Soru 1 — Doldur:** durum kodunu yorumlamalı; `anahtar yanlış` yazmalı.

```python
def durum_yorumla(kod):
    if kod == 200:
        return "tamam"
    elif kod == BOSLUK:
        return "anahtar yanlış"
    elif kod == 429:
        return "çok hızlı istek, biraz bekle"
    else:
        return "başka bir sorun"

print(durum_yorumla(401))
```

```python
geri = base64.BOSLUK(metin)
print(geri == baytlar)
```

**Soru 3 — Doldur:** baytları `kopya2.jpg` dosyasına yazmalı; sonra dosyanın var olduğunu söylemeli: `True`.

```python
with open("kopya2.jpg", BOSLUK) as dosya:
    dosya.write(BOSLUK)
print(Path("kopya2.jpg").exists())
```

**Soru 4 — Doldur:** yönergenin ilk 13 karakterinden, boşlukları tireye çevirerek bir dosya adı kurmalı: `a-red-bicycle_7.jpg`.

```python
yonerge = "a red bicycle on a yellow wall"
tohum = 7
ozet = yonerge[:BOSLUK].replace(" ", BOSLUK)
print(f"{ozet}_{tohum}.jpg")
```

**Soru 5 — Doldur:** bir tarihi dosya adına uygun biçimde yazmalı: `2026-10-19_09-05`. (Küçük `m` ay, büyük `M` dakika.)
`datetime(yıl, ay, gün, saat, dakika)` elle bir tarih kurar; `datetime.now()` yerine sabit bir tarih kullanıyoruz ki çıktı herkeste aynı olsun.

```python
tarih = datetime(2026, 10, 19, 9, 5)
print(tarih.strftime(BOSLUK))
```

**Soru 7 — Doldur:** metni sayıya çevirmeyi denemeli; çevrilemezse çökmeden `Sayıya çevrilemedi` yazmalı.

**Ek soru — Doldur:** küçük bir `kaydet` fonksiyonu. Ad alınmamışsa yazıp yolu döndürmeli, alınmışsa `None` döndürmeli.
İlk çalıştırmada `alistirma_ciktilari/fincan_2.jpg`, ikincide `None` yazmalı.

```python
klasor = Path("alistirma_ciktilari")
klasor.mkdir(exist_ok=True)

def kaydet(resim, ad):
    yol = klasor / ad
    if yol.BOSLUK():
        return None
    with open(yol, BOSLUK) as dosya:
        dosya.write(resim)
    return yol

print(kaydet(baytlar, "fincan_2.jpg"))
```
